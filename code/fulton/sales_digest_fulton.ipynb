# Examining the Data
Both the fulton digest and sales data were already compiled, so we don't actually need to do any of the compiling work here.

**Tax Digest**: fulton_digest_withnonprofit.csv  
**Sales Data**: SALES_fulton.csv

```python
import os
import pandas as pd
```

```python
SALES_PATH = "/Users/tpeng/Library/CloudStorage/OneDrive-GeorgiaInstituteofTechnology/Housing and Urban Policy (HUP) Lab - Documents/Data Files/Raw Data/ATLSales/FINAL/SALES_fulton.csv"
DIGEST_PATH = "/Users/tpeng/Library/CloudStorage/OneDrive-GeorgiaInstituteofTechnology/Housing and Urban Policy (HUP) Lab - Documents/Data Files/Final Datasets/Fulton/fulton_digest_withnonprofit.csv"

if os.path.exists(SALES_PATH):
    print(f"Sales file present at {SALES_PATH}")
else:
    print(f"Sales file NOT PRESENT, {SALES_PATH} not found.")

if os.path.exists(SALES_PATH):
    print(f"Digest file present at {DIGEST_PATH}")
else:
    print(f"Digest file NOT PRESENT, {DIGEST_PATH} not found.")
```

```python
digest_df = pd.read_csv(DIGEST_PATH, low_memory=False)
```

```python
digest_df.info()
```

```python
sales_df = pd.read_csv(SALES_PATH, low_memory=False)
```

```python
sales_df.info()
```

```python
OUT_PATH = "../../data/fulton/out"
```

```python
sales_tax_merged = pd.merge(left=sales_df, right=digest_df, left_on=["parcel_id", "tax_year"], right_on=["parid", "taxyr"],
                            suffixes=("_sales", ""))
```

```python
total_rows = sales_tax_merged.shape[0]
merged_rows = sales_tax_merged["parid"].notna().sum()

print(f"{merged_rows / total_rows} succesfully merged.")
```

```python
sales_tax_merged.to_csv(os.path.join(OUT_PATH, "FULTON_SALES_TAX_DIGEST_FINAL.csv"))
```
