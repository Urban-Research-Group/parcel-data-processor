## Determine which ParcelIDs are present and which are not
We want to check to see if the lack of a match is due only to the lack of a match in ParcelID, or specifically a lack of match (year, ParcelID) pairs.

```python
unmerged_pids = pd.Series(unmerged_df["parcel_id"].unique())
print(f"{unmerged_pids.size} unmerged unique parcelIDs")
```

```python
not_present = unmerged_pids[~unmerged_pids.isin(digest_df["parid"])]
present = unmerged_pids[unmerged_pids.isin(digest_df["parid"])]
print(f"{not_present.size} unmerged parcelIDs not present in digest")
```

```python
unmerged_not_present_df = unmerged_df[unmerged_df["parcel_id"].isin(not_present)]
unmerged_present_df = unmerged_df[unmerged_df["parcel_id"].isin(present)]
```

```python
unmerged_not_present_df.shape
```

```python
unmerged_present_df["year_before"] = unmerged_present_df["tax_year"] - 1
unmerged_present_df["year_after"] = unmerged_present_df["tax_year"] + 1

unmerged_year_status_df = pd.merge(unmerged_present_df, digest_df[["parid", "taxyr"]], left_on=["parcel_id", "year_before"], right_on=["parid", "taxyr"], how="left", suffixes=("", "_before"))
unmerged_year_status_df = pd.merge(unmerged_year_status_df, digest_df[["parid", "taxyr"]], left_on=["parcel_id", "year_after"], right_on=["parid", "taxyr"], how="left", suffixes=("", "_after"))
```

```python
before = (~unmerged_year_status_df.parid_before.isna()).sum()
after = (~unmerged_year_status_df.parid_after.isna()).sum()
both = ((~unmerged_year_status_df.parid_before.isna()) & (~unmerged_year_status_df.parid_after.isna())).sum()

print(f"{before}/{unmerged_present_df.shape[0]} parcels existed before sale\n{after}/{unmerged_present_df.shape[0]} parcels existed after \n{both}/{unmerged_present_df.shape[0]} parcels existed both before and after sale.\n" + 
      f"{unmerged_present_df.shape[0] - max(before, after, both)}/{unmerged_present_df.shape[0]} parcels existed neither before nor after.")
```

```python
unmerged_not_present_df.to_csv(os.path.join(OUT_PATH, "FULTON_UNMATCHED_NOT_PRESENT.csv"), index=False)
unmerged_year_status_df.to_csv(os.path.join(OUT_PATH, "FULTON_UNMATCHED_PRESENT.csv"), index=False)
```
