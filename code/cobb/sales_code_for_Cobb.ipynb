```python
import pandas as pd

# 1️ base on Cobb County layout
colspecs = [
    (0,30),   # PARID
    (30,39),  # TRANSNO
    (39,48),  # TRANSDT
    (48,56),  # BOOK
    (56,64),  # PAGE
    (64,73),  # SALEDT
    (73,84),  # PRICE
    (84,85),  # SOURCE
    (85,86),  # SALETYPE
    (86,88),  # SALEVAL
    (88,100)  # FILLER
]

colnames = [
    "PARID", "TRANSNO", "TRANSDT", "BOOK", "PAGE", "SALEDT",
    "PRICE", "SOURCE", "SALETYPE", "SALEVAL", "FILLER"
]

# read .DAT file
df = pd.read_fwf('/content/SALES.DAT', colspecs=colspecs, names=colnames)

# clean data
# delete filler
df = df.drop(columns=["FILLER"])

# Convert date columns to standard format (e.g. 29-MAR-19 → 2019-03-29)
for col in ["TRANSDT", "SALEDT"]:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Make sure PRICE is a numeric value
df["PRICE"] = pd.to_numeric(df["PRICE"], errors='coerce')

# Save as a CSV file
df.to_csv('/content/SALES_converted.csv', index=False)

print("Conversion complete! File saved as /content/SALES_converted.csv")
```
